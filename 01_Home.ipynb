**Alcohol consumption distribution**

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['alcperpeson'], kde=True, color='teal', bins=20)
plt.title('Distribution of Annual Alcohol Consumption per Person (Liters)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Alcohol Consumption (Liters)', fontsize=12)
plt.ylabel('Frequency (Country Count)', fontsize=12)
plt.tight_layout()
plt.show()
```

**Top 10 consuming countries**

```python
top_10 = df.nlargest(10, 'alcperpeson')

plt.figure(figsize=(12, 6))
barplot = sns.barplot(data=top_10, x='country', y='alcperpeson', palette='viridis')
plt.title('Top 10 Countries by Alcohol Consumption per Person', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Country', fontsize=12)
plt.ylabel('Alcohol Consumption (Liters)', fontsize=12)
plt.xticks(rotation=30)

for p in barplot.patches:
    barplot.annotate(f"{p.get_height():.2f}",
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='center',
                     xytext=(0, 5), textcoords='offset points')

plt.tight_layout()
plt.show()
```

**GDP vs Beverage price index**

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='gdppercapita', y='beveragepriceindex', hue='religion', palette='Set1', s=80, alpha=0.8)
plt.title('GDP per Capita vs. Beverage Price Index (Colored by Religion Factor)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('GDP per Capita (USD)', fontsize=12)
plt.ylabel('Beverage Price Index', fontsize=12)
plt.legend(title='Religion Factor', loc='upper right')
plt.tight_layout()
plt.show()
```

### 📊 Executive Summary & Key Performance Indicators (KPIs)
- The metrics below summarize baseline economic and consumption indicators across all analyzed countries.

```python
# Calculate key summary metrics
total_countries = df.shape[0]
avg_alc = df['alcperpeson'].mean()
avg_gdp = df['gdppercapita'].mean()
max_gdp_country = df.loc[df['gdppercapita'].idxmax(), 'country']
max_alc_country = df.loc[df['alcperpeson'].idxmax(), 'country']

# Display as a styled summary DataFrame or text block
kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Countries Analyzed",
        "Global Average Annual Alcohol Consumption",
        "Global Average GDP per Capita",
        "Highest Alcohol Consumption Country",
        "Highest GDP per Capita Country"
    ],
    "Value": [
        f"{total_countries}",
        f"{avg_alc:.2f} liters/person",
        f"${avg_gdp:,.2f}",
        f"{max_alc_country} ({df['alcperpeson'].max():.1f} L)",
        f"{max_gdp_country} (${df['gdppercapita'].max():,.2f})"
    ]
})

display(kpi_summary)
```

#### 🔗 Macro-Level Factor Correlations
- A high-level correlation matrix preview showing initial relationships between economic wealth, geography, and consumption.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
corr_matrix = df.drop(columns=['country']).corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, cbar_kws={'label': 'Correlation Coefficient'})
plt.title('Macro Correlation Overview', fontsize=13, fontweight='bold', pad=12)
plt.tight_layout()
plt.show()
```
